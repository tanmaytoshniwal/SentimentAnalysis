#### Helper function for concatenating string.

```
def give_list(para):
  arti=[]
  for _ in para:
    content=_.text
    arti.append(content)
  string=''.join(arti)
  give_res(string)
```

### Functions for scraping data from web.

```
def one(soup):
  results=soup.find('div',class_='article-entry')
  para=results.find_all('p')
  give_list(para)
```

```
def twothree(soup):
  results=soup.find('div',class_='et_pb_text_inner')
  par=results.find_all('p')
  para=results.find_all('li')
  arti=[]
  for _ in par:
    content=_.text
    arti.append(content)
  for _ in para:
    content=_.text
    arti.append(content)
  string=''.join(arti)
  give_res(string)
```

```
def four(soup,url):
  results=soup.find('article',class_='article article-regular post-118510 post type-post status-publish format-standard has-post-thumbnail hentry category-business tag-advice tag-book-excerpt tag-cleo-wade tag-how-to-be-a-better-human tag-passion')
  para=soup.select('p > *')
  #give_list(para)
  download_html(url)
```

```
def five(soup):
  results=soup.find('section',class_='entry')
  para=results.find_all('p')
  arti=[]
  for _ in para:
    content=_.text
    arti.append(content)
  arti1=arti[:12]
  string=''.join(arti1)
  give_res(string)
```

### Function for scraping data from web(using api).

```
def download_html(url):
  username = 'helloomannn'
  apiKey = '6er7ggWmdk6TZ1Sou9DgDoEJH'
  apiUrl = "http://api.scraping-bot.io/scrape/raw-html"
  payload = json.dumps({"url":url})
  headers = {
    'Content-Type': "application/json"
    }
  response = requests.request("POST", apiUrl, data=payload, auth=(username,apiKey), headers=headers)
  html_file = open("fetched.html","w")
  html_file.write(response.text)
  html_file.close()
  html_file = open("fetched.html","r")
  soup = BeautifulSoup(html_file)
  para=soup.find_all('p')
  give_list(para)
  html_file.close()
  os.remove("/content/fetched.html")
```

```
final=[]
f = open("/content/blogs_file_1.txt", "r")
count=0
for x in f:
  count+=1
  URL=x
  page=requests.get(URL)
  soup = BeautifulSoup(page.content, 'html.parser')
  if count==1:
    one(soup)
  if count==2:
    twothree(soup)
  if count==3:
    twothree(soup)
  if count==4:
    four(soup,URL)
  if count==5:
    five(soup)
f.close()

g = open("/content/blogs_file_2.txt", "r")
for x in g:
  download_html(x)
g.close()
```
